```python
import pandas as pd    
from sklearn.neighbors import KNeighborsClassifier   
from sklearn.preprocessing import StandardScaler 
from sklearn.model_selection import train_test_split 
from sklearn.metrics import accuracy_score, f1_score,roc_auc_score
```

```python
train = pd.read_csv("train.csv")
train.head(10)
```

```python
test = pd.read_csv("test.csv")
test.head(10)
```

```python
test_id = test["id"]
X_train = train.drop(columns=["id","diagnosed_diabetes"])
y = train["diagnosed_diabetes"]
X_test = test.drop("id",axis=1)
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X_train,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

X_train,X_val,y_train,y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.2,
    stratify=y_train,
    random_state=42
)
```

```python
X_train["bmi_whr"] = X_train["bmi"] * X_train["waist_to_hip_ratio"]
X_test["bmi_whr"] = X_test["bmi"] * X_test["waist_to_hip_ratio"]
X_val["bmi_whr"] = X_val["bmi"] * X_val["waist_to_hip_ratio"]
```

```python
features = X_train.select_dtypes(include="object").columns.tolist()

X_train = pd.get_dummies(X_train,columns=features,drop_first=False)
X_test = pd.get_dummies(X_test,columns=features,drop_first=False)
X_val = pd.get_dummies(X_val,columns=features,drop_first=False)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
X_val = scaler.transform(X_val)
```

```python
model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train,y_train)
```

```python
v_pred = model.predict(X_val)
acc = accuracy_score(y_val,v_pred)
print(acc)

f1 = f1_score(y_val,v_pred)
print(f1)

v_prob = model.predict_proba(X_val)[:, 1]
roc = roc_auc_score(y_val, v_prob)
print(roc)
```

```python
pred = model.predict(X_test)
acc = accuracy_score(y_test,pred)
print(acc)

f1 = f1_score(y_test,pred)
print(f1)


prob = model.predict_proba(X_test)[:, 1]
roc = roc_auc_score(y_test, prob)
print(roc)
```
